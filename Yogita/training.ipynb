```python
import os
import cv2
import torch
import torch.nn.functional as F
import numpy as np
from pathlib import Path
from psdc_sr_v3 import PSDCSRModel

# -------------------------------
# CONFIG
# -------------------------------
CKPT_PATH    = "/content/drive/MyDrive/CV_Project/checkpoints/ckpt_step_13400.pth"
DATASET_ROOT = "/content/drive/MyDrive/CV_Project/realsr/RealSR"
OUTPUT_ROOT  = "/content/drive/MyDrive/CV_Project/inference_results"

MODEL_SCALE   = 4
DEVICE        = "cuda" if torch.cuda.is_available() else "cpu"

# ✅ ONLY SCALE 4 (IMPORTANT FIX)
CAMERAS = ["Canon", "Nikon"]
SCALES  = ["4"]

# Optional (disable tiling for safety)
USE_TILING   = False
TILE_SIZE    = 256
TILE_OVERLAP = 16

# -------------------------------
# LOAD MODEL
# -------------------------------
def load_model():
    print(f"Loading checkpoint: {CKPT_PATH}")

    model = PSDCSRModel(scale=4, base_ch=32, latent_ch=16)

    ckpt = torch.load(CKPT_PATH, map_location=DEVICE)
    model.load_state_dict(ckpt['model_state'])

    model = model.to(DEVICE).eval()
    print(f"Loaded successfully on {DEVICE}\n")

    return model

# -------------------------------
# IMAGE IO
# -------------------------------
def read_image(path):
    img = cv2.imread(path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
    return torch.from_numpy(img).permute(2, 0, 1).unsqueeze(0)

def save_image(tensor, path):
    t = tensor.squeeze(0).permute(1, 2, 0).cpu().numpy()
    t = np.clip(t, 0, 1)
    t = (t * 255).astype(np.uint8)
    t = cv2.cvtColor(t, cv2.COLOR_RGB2BGR)

    os.makedirs(os.path.dirname(path), exist_ok=True)
    cv2.imwrite(path, t)

# -------------------------------
# MODEL FORWARD
# -------------------------------
@torch.no_grad()
def run_model(model, lr):
    lr = lr.to(DEVICE)

    # dummy HR (correct size)
    hr_dummy = F.interpolate(lr, scale_factor=model.scale,
                             mode="bicubic", align_corners=False)

    out = model(lr, lr, hr_dummy)

    # ensemble
    sr = (out["sr1"] + out["sr2"] + out["sr_self"]) / 3.0
    return sr.cpu()

# -------------------------------
# COLLECT FILES
# -------------------------------
def collect_pairs():
    pairs = []
    exts = {".png", ".jpg", ".jpeg"}

    for cam in CAMERAS:
        for scale in SCALES:
            folder = os.path.join(DATASET_ROOT, cam, "Test", scale)

            for fname in sorted(os.listdir(folder)):
                if Path(fname).suffix.lower() not in exts:
                    continue
                if "_HR" in fname:
                    continue

                src = os.path.join(folder, fname)
                dst = os.path.join(
                    OUTPUT_ROOT,
                    cam, "Test", scale,
                    Path(fname).stem + "_SR.png"
                )
                pairs.append((src, dst))

    return pairs

# -------------------------------
# MAIN
# -------------------------------
def main():
    print("\n===== PSDC-SR INFERENCE (Scale x4 ONLY) =====\n")

    model = load_model()
    pairs = collect_pairs()

    print(f"Found {len(pairs)} images\n")

    for i, (src, dst) in enumerate(pairs, 1):
        print(f"[{i}/{len(pairs)}] {src}")

        try:
            lr = read_image(src)

            sr = run_model(model, lr)

            save_image(sr, dst)

        except Exception as e:
            print("ERROR:", e)

    print("\n✅ DONE! Results saved at:")
    print(OUTPUT_ROOT)


if __name__ == "__main__":
    main()
```

```python
import cv2
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F

# pick 4 images
names = [
    "Canon_001_LR4",
    "Canon_002_LR4",
    "Canon_003_LR4",
    "Canon_004_LR4"
]

plt.figure(figsize=(12, 12))

for i, name in enumerate(names):

    # paths
    lr_path = f"/content/drive/MyDrive/CV_Project/realsr/RealSR/Canon/Test/4/{name}.png"
    sr_path = f"/content/drive/MyDrive/CV_Project/inference_results/Canon/Test/4/{name}_SR.png"

    # read images
    lr = cv2.cvtColor(cv2.imread(lr_path), cv2.COLOR_BGR2RGB)
    sr = cv2.cvtColor(cv2.imread(sr_path), cv2.COLOR_BGR2RGB)

    # bicubic upsample
    lr_tensor = torch.from_numpy(lr).permute(2,0,1).unsqueeze(0).float()/255
    bicubic = F.interpolate(lr_tensor, scale_factor=4, mode='bicubic', align_corners=False)
    bicubic = bicubic.squeeze().permute(1,2,0).numpy()

    # ---- plotting ----
    # LR
    plt.subplot(4, 3, i*3 + 1)
    plt.imshow(lr)
    plt.title(f"{name} - LR")
    plt.axis("off")

    # Bicubic
    plt.subplot(4, 3, i*3 + 2)
    plt.imshow(bicubic)
    plt.title("Bicubic")
    plt.axis("off")

    # SR
    plt.subplot(4, 3, i*3 + 3)
    plt.imshow(sr)
    plt.title("Model SR")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
